# OpenCV Basics

```python
# Imports
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Function to show the image
def show(img, title="Image"):
    plt.figure(figsize=(5,5))
    if len(img.shape) == 2:  
        # grayscale image
        plt.imshow(img, cmap="gray")
    else:
        plt.imshow(img)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

Read Image

```python
img = cv2.imread("image.png")
show(img, "Original Image")
```

Save Image

```python
cv2.imwrite("output.jpg", img)
```

Image Properties

```python
print("Shape:", img.shape)   # (height, width, channels)
print("Size:", img.size)
print("Dtype:", img.dtype)
```

Resize

```python
resized = cv2.resize(img, (300, 300))
show(resized, "Resized Image")
```

Color Conversion

```python
# BGR to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
show(rgb, "RGB Image")
```

```python
# BGR to Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
show(gray, "Grayscale Image")
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
show(hsv, "HSV Image")
```

Image Cropping

```python
cropped = img[50:250, 100:300]
show(cropped, "Cropped Image")
```

### Drawing Shapes
Line

```python
img_line = img.copy()
cv2.line(img_line, (50,50), (300,300), (0,255,0), 3)
show(img_line, "Line")
```

Rectangle

```python
img_rect = img.copy()
cv2.rectangle(img_rect, (50,50), (300,300), (255,0,0), 3)
show(img_rect, "Rectangle")
```

circle

```python
img_circle = img.copy()
cv2.circle(img_circle, (200,200), 80, (0,0,255), 3)
show(img_circle, "Circle")
```

Text

```python
img_text = img.copy()
cv2.putText(img_text, "OpenCV", (50,50),
            cv2.FONT_HERSHEY_SIMPLEX, 1, (255,255,255), 2)
show(img_text, "Text")
```

Thresholding

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
show(thresh, "Binary Threshold")
```

```python
show(gray, "Grayscale")
```

### Bluescreen replacement

```python
bluescreen_img = cv2.imread("bluescreen_tree.png")
# bluescreen_img = cv2.cvtColor(bluescreen_img, cv2.COLOR_BGR2RGB)
show(bluescreen_img, "Blue Screen")
```

```python
# Load the background image
background_img = cv2.imread("background.png")
background_img = cv2.cvtColor(background_img, cv2.COLOR_BGR2RGB)
show(background_img, "Background")
```

```python
# Set thresholds for blue color
lower_blue = np.array([0, 0, 100])     ##[R value, G value, B value]
upper_blue = np.array([0, 0, 255]) 
# upper_blue = np.array([120, 100, 255]) 
```

```python
# Create a mask for blue color
mask = cv2.inRange(bluescreen_img, lower_blue, upper_blue)
show(mask)
```

```python
# Create a mask image
masked_image = np.copy(bluescreen_img)
masked_image[mask != 0] = [0, 0, 0]
show(masked_image)
```

```python
bluescreen_img.shape, background_img.shape
```

```python
# Add background to the masked image

crop_background = background_img[0:720, 0:1280]
# Q. Instead of crop do the resizing

crop_background[mask == 0] = [0, 0, 0]

show(crop_background)
```

```python
final_image = crop_background + masked_image
show(final_image, "Final Image")
```

## Normalization

a. Min-max scaling

```python
# Min-max scaling
minmax_image = cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX)
show(minmax_image, "Final Image with Min-Max Scaling")
```

```python
# Z score normalization
mean = np.mean(img)
std = np.std(img)
zscore_image = (img - mean) / std
show(zscore_image, "Final Image with Z-Score Normalization")
```

```python
# Histogram equalization
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
equalized = cv2.equalizeHist(gray)
show(equalized, "Histogram Equalization")
```
