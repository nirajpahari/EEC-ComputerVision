## Data Augmentation

```python
# Imports
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Function to show the image
def show(img, title="Image"):
    plt.figure(figsize=(5,5))
    plt.imshow(img)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
img = cv2.imread("image.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

show(img, "Original Image")
```

### 1 Geometric Transformation
#### Rotation

```python
(h, w) = img.shape[:2]
center = (w // 2, h // 2)

M = cv2.getRotationMatrix2D(center, 30, 1.0)
rotated = cv2.warpAffine(img, M, (w, h))

show(rotated, "Rotated")
```

#### Flipping

```python
horizontal_flip = cv2.flip(img, 1)
vertical_flip = cv2.flip(img, 0)

show(horizontal_flip, "Horizontal Flip")
show(vertical_flip, "Vertical Flip")
```

#### Scaling

```python
scaled = cv2.resize(img, (200, 250))
# scaled = cv2.resize(img, None, fx=1.2, fy=1.2)
show(scaled, "Scaled")
```

#### Translation

```python
(h, w) = img.shape[:2]
M = np.float32([[1, 0, 50],
                [0, 1, 30]])

translated = cv2.warpAffine(img, M, (w, h))
show(translated, "Translated")
```

#### Cropping

```python
cropped = img[50:300, 80:350]
show(cropped, "Cropped")
```

### 2. Color Space Augmentation
#### Brightness

```python
bright = cv2.convertScaleAbs(img, alpha=1.2, beta=30)
show(bright, "Brightness Adjusted")
```

#### Contrast

```python
contrast = cv2.convertScaleAbs(img, alpha=1.5, beta=0)
show(contrast, "Contrast Adjusted")
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV).astype(np.float32)

hsv[..., 0] = (hsv[..., 0] + 10) % 180 # Shift hue by 10 degrees
hsv[..., 1] = np.clip(hsv[..., 1] * 1.2, 0, 255) # Increase saturation by 20%
hsv[..., 2] = np.clip(hsv[..., 2] * 1.1, 0, 255) # Increase value by 10%

hsv = hsv.astype(np.uint8)
color_shifted = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

show(color_shifted, "Hue & Saturation Shift")
```

### 3. Noise Injection
#### Gaussian Noise

```python
noise = np.random.normal(0, 10, img.shape)
noisy = np.clip(img + noise, 0, 255).astype(np.uint8)

show(noisy, "Gaussian Noise")
```

#### Salt & Peeper Noise

```python
sp = img.copy()
prob = 0.01

rand = np.random.rand(*img.shape[:2])

sp[rand < prob] = 0
sp[rand > 1 - prob] = 255

show(sp, "Salt & Pepper Noise")
```

### 4. Blurring and Sharpening
#### Gaussian Blur

```python
blur = cv2.GaussianBlur(img, (5,5), 0)
show(blur, "Gaussian Blur")
```

```python
kernel = np.array([[0,-1,0],
                   [-1,5,-1],
                   [0,-1,0]])

sharp = cv2.filter2D(img, -1, kernel)
show(sharp, "Sharpened")
```
